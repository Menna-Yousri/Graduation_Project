```python
# ==================== SETUP ====================
!pip install -q ultralytics
import ultralytics
ultralytics.checks()

from ultralytics import YOLO
import os
import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

# ==================== MOUNT GOOGLE DRIVE ====================
from google.colab import drive
drive.mount('/content/drive')

# ==================== SET PATHS ====================
base_dir = "/content/drive/MyDrive/MooNet Datasets/Cow_Detection_Dataset/Cow_Data"
images_dir = os.path.join(base_dir, 'images')
labels_dir = os.path.join(base_dir, 'labels')

# ==================== VISUALIZATION FUNCTION ====================
def visualize_bounding_boxes(image_path, bboxes):
    """
    Visualizes bounding boxes on an image using Matplotlib.
    :param image_path: Path to the image file.
    :param bboxes: List of bounding boxes in YOLO format: [x_center, y_center, width, height] (normalized).
    """
    image = cv2.cvtColor(cv2.imread(image_path), cv2.COLOR_BGR2RGB)
    height, width = image.shape[:2]
    fig, ax = plt.subplots(1)
    ax.imshow(image)

    for bbox in bboxes:
        x_center, y_center, w, h = bbox
        x_center *= width
        y_center *= height
        w *= width
        h *= height
        x = x_center - w / 2
        y = y_center - h / 2
        rect = patches.Rectangle((x, y), w, h, linewidth=2, edgecolor='g', facecolor='none')
        ax.add_patch(rect)

    plt.axis('off')
    plt.show()

# ==================== DISPLAY 6 RANDOM SAMPLES ====================
samples = np.random.choice(os.listdir(os.path.join(images_dir, 'train')), 6)
labels_path = os.path.join(labels_dir, 'train')

for sample in samples:
    print(f'sample = {sample}')
    image_path = os.path.join(images_dir, 'train', sample)
    label_path = os.path.join(labels_path, sample.replace('jpg', 'txt'))
    bboxes = []
    with open(label_path, 'r') as f:
        contents = f.read()
    for content in contents.strip().split('\n'):
        if len(content) > 1:
            bbx = content.split(' ')
            bboxes.append([float(bbx[1]), float(bbx[2]), float(bbx[3]), float(bbx[4])])
            print(bbx)
    visualize_bounding_boxes(image_path, bboxes)
    print('*******************************')

# ==================== DATASET STATS ====================
splits = ['train', 'validation', 'test']
total_images = 0
print("📊 Dataset Summary:\n")

for split in splits:
    image_split_path = os.path.join(images_dir, split)
    label_split_path = os.path.join(labels_dir, split)
    num_images = len([f for f in os.listdir(image_split_path) if f.endswith('.jpg')])
    num_labels = len([f for f in os.listdir(label_split_path) if f.endswith('.txt')])
    total_images += num_images
    print(f"Number of {split} images: {num_images}")
    print(f"Number of {split} labels: {num_labels}")
    print()

print(f"Total number of images: {total_images}")

# ==================== CREATE data.yaml ====================
data_yaml_path = os.path.join(base_dir, 'data.yaml')
with open(data_yaml_path, 'w') as f:
    f.write(f"""
train: {images_dir}/train
val: {images_dir}/validation
test: {images_dir}/test

nc: 1
names: ['cow']
""")
print(f"'data.yaml' created at: {data_yaml_path}")

# ==================== TRAIN YOLOv8 ====================
model = YOLO("yolov8n.pt")
model.train(
    data=data_yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    project=os.path.join(base_dir, 'runs'),
    name="yolov8-cow-detection"
)
```

```python
metrics = model.val()
print(metrics)
```

```python
test_images_path = os.path.join(images_dir, 'test')
test_samples = np.random.choice(os.listdir(test_images_path), 3)

for sample in test_samples:
    image_path = os.path.join(test_images_path, sample)
    results = model.predict(source=image_path, conf=0.25, save=False)

    result_img = results[0].plot()
    plt.figure(figsize=(8, 8))
    plt.imshow(result_img)
    plt.axis('off')
    plt.title(f"Prediction: {sample}")
    plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import math

# مسار مجلد الصور
metrics_dir = os.path.join(base_dir, 'runs', 'yolov8-cow-detection52')

if os.path.exists(metrics_dir):
    image_files = [f for f in os.listdir(metrics_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    if image_files:
        num_images = len(image_files)
        cols = 3  # عدد الأعمدة
        rows = math.ceil(num_images / cols)

        plt.figure(figsize=(15, 5 * rows))

        for i, img_name in enumerate(image_files):
            img_path = os.path.join(metrics_dir, img_name)
            img = mpimg.imread(img_path)

            plt.subplot(rows, cols, i + 1)
            plt.imshow(img)
            plt.axis('off')
            plt.title(img_name, fontsize=10)

        plt.tight_layout()
        plt.show()
    else:
        print("⚠️ No image files found in the metrics directory.")
else:
    print("⚠️ Metrics directory not found.")
```
